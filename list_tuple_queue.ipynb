```python
# basic operation
lst=["Python", "Java", "C++", "JavaScript"]
```

```python
length = len(lst)
print("Length of the list:", length)
```

```python
first_element = lst[0]
print("First element of the list:", first_element)
```

```python
last_element = lst[-1]
print("Last element of the list:", last_element)
```

```python
last_element = lst[3]
print("Last element of the list:", last_element)
```

```python
lst[2]='sql'
print("Updated list:", lst)
```

```python
lst.append("data bricks")
print("List after appending an element:", lst)
```

```python
lst.remove("Java")
print("List after removing an element:", lst)
```
